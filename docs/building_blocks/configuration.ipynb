# General configurations

This page provides a short overview on globally configurable settings. At the
moment, these are only concerning prints of components, connections and
properties but they will be extended in the future.

## Display modes

Connections, components and their data containers provide `repr` and `print`
output. You can control the level of detail globally with
`tespy.set_display_mode`. Available modes are:

- `"compact"` (default)
- `"extensive"`
- `"none"`

For example, consider the following model:

```python
import tespy
from tespy.components import MovingBoundaryHeatExchanger, Sink, Source
from tespy.connections import Connection, Ref
from tespy.networks import Network

nw = Network(iterinfo=False)
nw.units.set_defaults(
    pressure="bar", pressure_difference="bar", temperature="degC",
    enthalpy="kJ/kg", heat="kW"
)

cond = MovingBoundaryHeatExchanger("condenser")
c1 = Connection(Source("refrigerant inflow"), "out1", cond, "in1", label="1")
c2 = Connection(cond, "out1", Sink("refrigerant outflow"), "in1", label="2")
c11 = Connection(Source("water inflow"), "out1", cond, "in2", label="11")
c12 = Connection(cond, "out2", Sink("water outflow"), "in1", label="12")
nw.add_conns(c1, c2, c11, c12)

c1.set_attr(fluid={"PROPANE": 1}, T=70)
c2.set_attr(x=0)
c11.set_attr(fluid={"water": 1}, p=1, T=30)
c12.set_attr(T=Ref(c11, 1, 10))
cond.set_attr(Q=-15, td_pinch=3, pr1=1, pr2=1)
nw.solve("design")
```
